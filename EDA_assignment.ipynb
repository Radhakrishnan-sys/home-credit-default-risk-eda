### Univariate Analysis of the Categorical Data

````python
def plotting(train, train_0, train_1, column):
    fig = plt.figure(figsize=(13, 10))

    # Pie chart for overall distribution
    ax1 = plt.subplot(221)
    train[column].value_counts().plot.pie(autopct='%1.0f%%', ax=ax1)
    ax1.set_title('Pie Chart of ' + column)

    # Countplot for train_0
    ax2 = plt.subplot(222)
    sns.countplot(x=column, hue='TARGET', data=train_0, ax=ax2)
    ax2.set_title('Countplot of ' + column + ' for TARGET=0')

    # Bar plot for percentage comparison
    ax3 = plt.subplot(223)
    df = pd.DataFrame()
    df['0'] = (train_0[column].value_counts() / len(train_0)) * 100  # Percentage for TARGET=0
    df['1'] = (train_1[column].value_counts() / len(train_1)) * 100  # Percentage for TARGET=1
    df.plot.bar(ax=ax3)
    ax3.set_title('Percentage Comparison of ' + column + ' for TARGET=0 and TARGET=1')

    fig.tight_layout()
    plt.show()
````

````python
for column in train_categorical:
    print('Plotting', column)
    plotting(app, train_0, train_1,column )
    print('--------------------------------------------------------------------------------')
````
